**Problems found in those results (all handled in this v2 notebook)**

1. **GroupDRO collapsed onto a single hospital in all 18 runs** (`q` = [1, 0] or [0, 1], every seed, ViT and ConvNeXt). "GroupDRO" is therefore single-hospital training, and its apparent gains over ERM (balanced) come from ignoring the small VinBigData source. Part H2 compares it with the matching single-source model and removes it from the DG verdict.
2. **The label-free prior-shift correction failed** for NIH→CheXpert (ECE rose from about 0.11 to about 0.45, and the EM prevalence estimate was far worse than assuming no shift). The label-shift assumption does not hold. Part G3 says so automatically.
3. **Two definitions of the asymmetry were in use** (0.073 from seed-ensemble probabilities, 0.081 from per-seed means). v2 uses the per-seed mean everywhere and shows the ensemble version as a sensitivity analysis (Part L1).
4. **"% of the F1 gap recovered" was misleading** (values above 100% and about 0%), because F1 depends on prevalence and the two hospitals differ in prevalence. v2 normalizes by the F1 an oracle threshold would reach on the same ranking.
5. **No new significance claim from a 300-resample bootstrap.** Every p-value here uses NB = 2000.

```python
def fauc(y, s):
    y = np.asarray(y).astype(bool); r = rankdata(s); n1 = y.sum(); n0 = len(y) - n1
    return (r[y].sum() - n1 * (n1 + 1) / 2) / (n1 * n0)
def lab_auc(Y, P, valid):                                            # per-label AUROC, NaN where the label is absent or too rare
    return np.array([fauc(Y[:, j], P[:, j]) if valid[j] and 5 <= Y[:, j].sum() <= len(Y) - 5 else np.nan for j in range(L)])
def mauc(Y, P, valid): return float(np.nanmean(lab_auc(Y, P, valid)))
def mauprc(Y, P, valid): return float(np.mean([average_precision_score(Y[:, j], P[:, j]) for j in range(L) if valid[j] and Y[:, j].sum() >= 5]))
sig = lambda z: 1 / (1 + np.exp(-z))
def logit(p): p = np.clip(p, 1e-6, 1 - 1e-6); return np.log(p / (1 - p))
GRID = np.linspace(0.02, 0.95, 94)
def best_thr(y, p):                                                  # threshold that maximises F1 on (y, p)
    pr = p[:, None] >= GRID[None, :]; y1 = (y == 1)[:, None]; tp = (pr & y1).sum(0); fp = pr.sum(0) - tp; fn = y1.sum() - tp
    return GRID[np.argmax(2 * tp / np.maximum(2 * tp + fp + fn, 1))]
def f1_at(y, p, t):
    pr = p >= t; y1 = y == 1; tp = (pr & y1).sum(); d = 2 * tp + (pr & ~y1).sum() + (~pr & y1).sum(); return 2 * tp / d if d else 0.0
def ece_of(y, p, nb=10):
    bi = np.clip((p * nb).astype(int), 0, nb - 1); e = 0.0
    for k in range(nb):
        mk = bi == k
        if mk.any(): e += mk.mean() * abs(p[mk].mean() - y[mk].mean())
    return e
def reliab(y, p, nb=10):
    bi = np.clip((p * nb).astype(int), 0, nb - 1); xs_, ys_ = [], []
    for k in range(nb):
        mk = bi == k
        if mk.sum() > 30: xs_.append(p[mk].mean()); ys_.append(y[mk].mean())
    return xs_, ys_
def holm(p):                                                         # Holm step-down adjusted p-values
    p = np.asarray(p, float); o = np.argsort(p); out = np.empty(len(p)); run = 0.0
    for r, i in enumerate(o): run = max(run, min(1.0, (len(p) - r) * p[i])); out[i] = run
    return out
def boot_p(d, nb): return min(1.0, 2 * min((d <= 0).mean(), (d >= 0).mean()) + 1.0 / nb)    # two-sided bootstrap p; floor is 1/nb, so nb must be large for Holm
pci = lambda a: np.percentile(a, [2.5, 97.5], axis=0)
def groups(p): return list(pd.Series(np.arange(len(p))).groupby(np.asarray(p)).apply(np.array).values)
def resample(T, rng):                                                # patient-level bootstrap indices for test site T (image-level for C: no patient id)
    if GRP[T] is None: n = len(YT[T]); return rng.integers(0, n, n)
    g = GRP[T]; return np.concatenate([g[k] for k in rng.integers(0, len(g), len(g))])
dname = lambda ts, T: f"{ts}→{T}"
```

# Part B: models and training (skips finished runs)

* **ViT-S:** ERM (natural) on all six training sets, plus ERM (balanced), DANN, CORAL, GroupDRO on the three pooled sets.
* **ConvNeXt-T:** the four DG methods on the three pooled sets. ConvNeXt-T ERM (natural) and all DenseNet121 models come from your earlier runs (`fair_ens` logits) and are not retrained.
* **Recipes** follow each architecture's standard fine-tuning setup and are reported as such in the paper: DenseNet121 and ConvNeXt-T use AdamW, LR 1e-4, weight decay 1e-4, cosine schedule, no warmup, no gradient clipping. ViT-S uses LR 5e-5, weight decay 0.05, 300 warmup steps, gradient clipping 1.0 and drop-path 0.1.
* **DG settings** are identical to your DenseNet runs: DANN weight 0.1 with the lambda ramp, CORAL weight 1.0, GroupDRO eta 0.01. Model selection is always the mean source-validation AUROC over the training hospitals.
* **Known property of GroupDRO here:** it uses the raw group loss as its hardness measure. If one hospital has a higher label prevalence its loss is always larger, and the weights can collapse onto it (`q = [1, 0]` in your ConvNeXt B+C log). The final `q` weights are tabulated in Part H and must be reported.

```python
IMN_M, IMN_S = [0.485, 0.456, 0.406], [0.229, 0.224, 0.225]
try:
    _c = timm.create_model(ARCH_CFG["ViT-S"]["timm"], pretrained=False, num_classes=L).pretrained_cfg; _vm, _vs = list(_c["mean"]), list(_c["std"])
except Exception:
    _vm, _vs = [0.5] * 3, [0.5] * 3                                                  # augreg ViT weights use 0.5 / 0.5
NORM = {a: (torch.tensor(m_, device=dev).view(1, 3, 1, 1), torch.tensor(s_, device=dev).view(1, 3, 1, 1))
        for a, (m_, s_) in {"DenseNet121": (IMN_M, IMN_S), "ConvNeXt-T": (IMN_M, IMN_S), "ViT-S": (_vm, _vs)}.items()}
CUR = {"a": "ViT-S"}
def use_arch(a):
    CUR.clear(); CUR.update(ARCH_CFG[a]); CUR["a"] = a
    if ARCH_CFG[a]["dir"]: os.makedirs(ARCH_CFG[a]["dir"], exist_ok=True)
def make_model(pretrained=True, a=None):
    c = ARCH_CFG[a or CUR["a"]]; kw = dict(drop_path_rate=c["drop_path"]) if c["drop_path"] > 0 else {}
    return timm.create_model(c["timm"], pretrained=pretrained, num_classes=L, **kw).to(dev)
def gpu_batch(A, idx, flip=False, a=None):
    mu, sd = NORM[a or CUR["a"]]
    x = torch.from_numpy(np.ascontiguousarray(A[np.asarray(idx)])).to(dev).float().div(255).unsqueeze(1)
    if flip:
        f = torch.rand(len(idx), device=dev) < 0.5; x = torch.where(f.view(-1, 1, 1, 1), x.flip(-1), x)
    return (x.repeat(1, 3, 1, 1) - mu) / sd
@torch.inference_mode()
def predict(m, A, idx, bs=128, a=None):
    m.eval(); out = []
    for i in range(0, len(idx), bs):
        with torch.autocast("cuda", enabled=(dev == "cuda")): o = m(gpu_batch(A, idx[i:i + bs], a=a))
        out.append(torch.sigmoid(o.float()).cpu())
    return torch.cat(out).numpy()
def load_arch(a, path):
    m = make_model(False, a); m.load_state_dict(torch.load(path, map_location=dev, weights_only=True)); return m.eval()
def val_scores(m, srcs):                                                             # SOURCE validation only, never the target
    return {k: mauc(SRC[k]["Y"][SRC[k]["va"]], predict(m, SRC[k]["arr"], SRC[k]["va"]), SRC[k]["valid"]) for k in srcs}

def batches(srcs, seed, ep):                                                         # natural pooled sampling
    rng = np.random.default_rng(seed * 100 + ep)
    dom = np.concatenate([np.full(len(SRC[k]["tr"]), i) for i, k in enumerate(srcs)]); pos = np.concatenate([SRC[k]["tr"] for k in srcs])
    perm = rng.permutation(len(pos))
    for k in range(math.ceil(len(pos) / BS)):
        s = perm[k * BS:(k + 1) * BS]
        yield {srcs[i]: np.sort(pos[s][dom[s] == i]) for i in range(len(srcs)) if (dom[s] == i).any()}
def batches_bal(srcs, seed, ep):                                                     # equal number of images from every hospital in every step
    rng = np.random.default_rng(seed * 100 + ep); h = BS // len(srcs)
    for _ in range(math.ceil(sum(len(SRC[k]["tr"]) for k in srcs) / BS)):
        yield {k: np.sort(rng.choice(SRC[k]["tr"], h, replace=False)) for k in srcs}
class GRL(torch.autograd.Function):                                                  # gradient reversal for DANN
    @staticmethod
    def forward(ctx, x, lam): ctx.lam = lam; return x.view_as(x)
    @staticmethod
    def backward(ctx, g): return -ctx.lam * g, None
def coral_loss(fa, fb):                                                              # match the feature covariances of the two hospitals
    d = fa.size(1); return ((torch.cov(fa.T) - torch.cov(fb.T)) ** 2).sum() / (4 * d * d)
def finish_and_save(m, srcs, BEST, PRD):                                             # predictions of the best checkpoint on every clean test site
    m.load_state_dict(torch.load(BEST, map_location=dev, weights_only=True)); z = {}
    z["A_test_P"] = predict(m, NIH_A, te_i); z["B_test_P"] = predict(m, CX_T, np.arange(len(CX_T)))
    if "C" not in srcs: z["C_test_P"] = predict(m, VB_A, np.arange(NVB))             # C is a clean test site only if it was not trained on
    for k in ["A", "B"] + (["C"] if "C" in srcs else []): z[f"{k}_val_P"] = predict(m, SRC[k]["arr"], SRC[k]["va"])
    np.savez(PRD, **z)
def _paths(a, ts, me, seed):
    b = stem(a, ts, me, seed); return b + ".pt", b + "_last.pt", b + "_history.csv", b + "_preds.npz"
def _resume(LAST, m, opt, sch, scaler, disc=None):
    ck = torch.load(LAST, map_location=dev, weights_only=False)
    m.load_state_dict(ck["model"]); opt.load_state_dict(ck["opt"]); sch.load_state_dict(ck["sched"]); scaler.load_state_dict(ck["scaler"])
    if disc is not None and ck.get("disc") is not None: disc.load_state_dict(ck["disc"])
    print("resumed, next epoch", ck["ep"] + 2); return ck
```

```python
def train_vit_natural(ts, seed):                                                     # ViT-S ERM with natural pooled sampling
    a = "ViT-S"; use_arch(a); srcs = ts.split("+"); BEST, LAST, HIST, PRD = _paths(a, ts, NAT, seed)
    if os.path.exists(PRD): print("skip (finished):", os.path.basename(PRD)); return
    torch.manual_seed(seed); np.random.seed(seed)
    m = make_model(); opt = torch.optim.AdamW(m.parameters(), lr=CUR["lr"], weight_decay=CUR["wd"])
    spe = math.ceil(sum(len(SRC[k]["tr"]) for k in srcs) / BS); total, warm = EPOCHS * spe, min(300, spe)
    sch = torch.optim.lr_scheduler.LambdaLR(opt, lambda t: (t + 1) / warm if t < warm else 0.5 * (1 + math.cos(math.pi * (t - warm) / max(1, total - warm))))
    scaler = torch.amp.GradScaler("cuda", enabled=(dev == "cuda")); hist, best, start = [], 0.0, 0
    if os.path.exists(LAST):
        ck = _resume(LAST, m, opt, sch, scaler); hist, best, start = ck["hist"], ck["best"], ck["ep"] + 1
    for ep in range(start, EPOCHS):
        m.train(); t0 = time.time(); tl = torch.zeros((), device=dev); n = 0
        for batch in tqdm(batches(srcs, seed, ep), total=spe, leave=False, desc=f"ViT {ts} ERM(nat) seed {seed} ep {ep + 1}/{EPOCHS}"):
            xs, ys, ms = [], [], []
            for k, idx in batch.items():
                xs.append(gpu_batch(SRC[k]["arr"], idx, flip=True)); ys.append(SRC[k]["Y"][idx]); ms.append(np.tile(SRC[k]["valid"], (len(idx), 1)))
            x = torch.cat(xs); y = torch.from_numpy(np.concatenate(ys)).to(dev).float(); mk = torch.from_numpy(np.concatenate(ms)).to(dev).float()
            opt.zero_grad(set_to_none=True)
            with torch.autocast("cuda", enabled=(dev == "cuda")): z = m(x)
            loss = (F.binary_cross_entropy_with_logits(z.float(), y, reduction="none") * mk).sum() / mk.sum()      # masked BCE: Edema is ignored for VinBigData
            scaler.scale(loss).backward(); scaler.unscale_(opt); nn.utils.clip_grad_norm_(m.parameters(), 1.0)
            scaler.step(opt); scaler.update(); sch.step(); tl += loss.detach() * len(x); n += len(x)
        vs = val_scores(m, srcs); vm = float(np.mean(list(vs.values())))
        row = dict(epoch=ep + 1, train_loss=float(tl) / n, val_mean=vm, **{f"val_{k}": v for k, v in vs.items()}, sec=time.time() - t0)
        hist.append(row); pd.DataFrame(hist).to_csv(HIST, index=False)
        print(f"ViT {ts} ERM(nat) seed {seed} ep {ep + 1} | loss {row['train_loss']:.4f} | val AUROC {vs} | {row['sec'] / 60:.1f} min")
        if vm > best: best = vm; torch.save(m.state_dict(), BEST)
        torch.save(dict(model=m.state_dict(), opt=opt.state_dict(), sched=sch.state_dict(), scaler=scaler.state_dict(), hist=hist, best=best, ep=ep), LAST)
    finish_and_save(m, srcs, BEST, PRD); del m, opt, sch, scaler; gc.collect(); torch.cuda.empty_cache()

def train_dg_vit(ts, meth, seed, lam_coral=1.0, lam_dann=0.1, eta_dro=0.01):         # ViT-S with ERM (balanced), DANN, CORAL or GroupDRO
    a = "ViT-S"; use_arch(a); assert meth != NAT and len(ts.split("+")) == 2
    srcs = ts.split("+"); BEST, LAST, HIST, PRD = _paths(a, ts, meth, seed)
    if os.path.exists(PRD): print("skip (finished):", os.path.basename(PRD)); return
    torch.manual_seed(seed); np.random.seed(seed)
    m = make_model(); disc = nn.Sequential(nn.Linear(m.num_features, 256), nn.ReLU(), nn.Linear(256, 2)).to(dev) if meth == "DANN" else None
    opt = torch.optim.AdamW(list(m.parameters()) + (list(disc.parameters()) if disc else []), lr=CUR["lr"], weight_decay=CUR["wd"])
    spe = math.ceil(sum(len(SRC[k]["tr"]) for k in srcs) / BS); total, warm = EPOCHS * spe, min(300, spe)
    sch = torch.optim.lr_scheduler.LambdaLR(opt, lambda t: (t + 1) / warm if t < warm else 0.5 * (1 + math.cos(math.pi * (t - warm) / max(1, total - warm))))
    scaler = torch.amp.GradScaler("cuda", enabled=(dev == "cuda")); hist, best, start = [], 0.0, 0; q = torch.ones(2, device=dev) / 2
    if os.path.exists(LAST):
        ck = _resume(LAST, m, opt, sch, scaler, disc); hist, best, start, q = ck["hist"], ck["best"], ck["ep"] + 1, ck["q"].to(dev)
    for ep in range(start, EPOCHS):
        m.train(); t0 = time.time(); tl = torch.zeros((), device=dev); n = 0
        for step, batch in enumerate(tqdm(batches_bal(srcs, seed, ep), total=spe, leave=False, desc=f"ViT {ts} {meth} seed {seed} ep {ep + 1}/{EPOCHS}")):
            x = torch.cat([gpu_batch(SRC[k]["arr"], batch[k], flip=True) for k in srcs]); sizes = [len(batch[k]) for k in srcs]
            ys = [torch.from_numpy(SRC[k]["Y"][batch[k]]).to(dev).float() for k in srcs]
            mk = [torch.from_numpy(np.tile(SRC[k]["valid"], (len(batch[k]), 1))).to(dev).float() for k in srcs]
            opt.zero_grad(set_to_none=True)
            with torch.autocast("cuda", enabled=(dev == "cuda")):
                f = m.forward_head(m.forward_features(x), pre_logits=True); z = m.get_classifier()(f)
            z, f = z.float(), f.float(); zs, fs = torch.split(z, sizes), torch.split(f, sizes)
            gsum = [(F.binary_cross_entropy_with_logits(zz, yy, reduction="none") * mm).sum() for zz, yy, mm in zip(zs, ys, mk)]; gden = [mm.sum() for mm in mk]
            if meth == "GroupDRO":                                                   # exponentiated-gradient weights on the per-hospital losses
                gl = torch.stack([g_ / d_ for g_, d_ in zip(gsum, gden)]); q = q * torch.exp(eta_dro * gl.detach()); q = q / q.sum(); loss = (q * gl).sum()
            else:
                loss = sum(gsum) / sum(gden)                                         # masked pooled BCE (same as ERM)
                if meth == "CORAL": loss = loss + lam_coral * coral_loss(fs[0], fs[1])
                if meth == "DANN":                                                   # 0.1 x domain loss with the lambda ramp, as in your DenseNet runs
                    ramp = 2 / (1 + math.exp(-10 * (ep * spe + step) / total)) - 1
                    dom = torch.cat([torch.full((s_,), i, device=dev) for i, s_ in enumerate(sizes)])
                    loss = loss + lam_dann * F.cross_entropy(disc(GRL.apply(f, ramp)), dom)
            scaler.scale(loss).backward(); scaler.unscale_(opt); nn.utils.clip_grad_norm_(m.parameters(), 1.0)
            scaler.step(opt); scaler.update(); sch.step(); tl += loss.detach() * len(x); n += len(x)
        vs = val_scores(m, srcs); vm = float(np.mean(list(vs.values())))
        row = dict(epoch=ep + 1, train_loss=float(tl) / n, val_mean=vm, **{f"val_{k}": v for k, v in vs.items()}, sec=time.time() - t0)
        hist.append(row); pd.DataFrame(hist).to_csv(HIST, index=False)
        print(f"ViT {ts} {meth} seed {seed} ep {ep + 1} | loss {row['train_loss']:.4f} | val {vs}" + (f" | q={q.cpu().numpy().round(2)}" if meth == "GroupDRO" else ""))
        if vm > best: best = vm; torch.save(m.state_dict(), BEST)
        torch.save(dict(model=m.state_dict(), opt=opt.state_dict(), sched=sch.state_dict(), scaler=scaler.state_dict(), hist=hist, best=best, ep=ep, q=q.cpu(), **({"disc": disc.state_dict()} if disc else {})), LAST)
    finish_and_save(m, srcs, BEST, PRD); del m, opt, sch, scaler; gc.collect(); torch.cuda.empty_cache()

def train_dg_cnx(ts, meth, seed, dann_w=0.1, coral_w=1.0, dro_eta=0.01):             # ConvNeXt-T, recipe identical to your DenseNet runs
    a = "ConvNeXt-T"; use_arch(a); assert meth != NAT and len(ts.split("+")) == 2
    srcs = ts.split("+"); BEST, LAST, HIST, PRD = _paths(a, ts, meth, seed)
    if os.path.exists(PRD): print("skip (finished):", os.path.basename(PRD)); return
    torch.manual_seed(seed); np.random.seed(seed)
    m = make_model(); disc = nn.Sequential(nn.Linear(m.num_features, 256), nn.ReLU(), nn.Linear(256, len(srcs))).to(dev) if meth == "DANN" else None
    opt = torch.optim.AdamW(list(m.parameters()) + (list(disc.parameters()) if disc else []), lr=CUR["lr"], weight_decay=CUR["wd"])
    spe = math.ceil(sum(len(SRC[k]["tr"]) for k in srcs) / BS); sch = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=EPOCHS * spe)
    scaler = torch.amp.GradScaler("cuda", enabled=(dev == "cuda")); hist, best, start = [], 0.0, 0; q = torch.ones(len(srcs), device=dev) / len(srcs)
    if os.path.exists(LAST):
        ck = _resume(LAST, m, opt, sch, scaler, disc); hist, best, start, q = ck["hist"], ck["best"], ck["ep"] + 1, ck["q"].to(dev)
    tt = lambda arrs, dt: torch.from_numpy(np.concatenate(arrs)).to(dev).to(dt)
    for ep in range(start, EPOCHS):
        m.train(); t0 = time.time(); tl = torch.zeros((), device=dev)
        for k, batch in enumerate(tqdm(batches_bal(srcs, seed, ep), total=spe, leave=False, desc=f"ConvNeXt {ts} {meth} seed {seed} ep {ep + 1}/{EPOCHS}")):
            xs, ys, ms, ds = [], [], [], []
            for i, s_ in enumerate(srcs):
                idx = batch[s_]; xs.append(gpu_batch(SRC[s_]["arr"], idx, flip=True)); ys.append(SRC[s_]["Y"][idx])
                ms.append(np.tile(SRC[s_]["valid"], (len(idx), 1))); ds.append(np.full(len(idx), i))
            x = torch.cat(xs); y, mk, d = tt(ys, torch.float32), tt(ms, torch.float32), tt(ds, torch.long)
            with torch.autocast("cuda", enabled=(dev == "cuda")):
                f = m.forward_head(m.forward_features(x), pre_logits=True); z = m.get_classifier()(f)
            f = f.float(); ls = (F.binary_cross_entropy_with_logits(z.float(), y, reduction="none") * mk).sum(1) / mk.sum(1).clamp(min=1)   # per-sample loss, Edema masked for C
            if meth == "ERM (balanced)": loss = ls.mean()
            elif meth == "GroupDRO":
                gl = torch.stack([ls[d == i].mean() for i in range(len(srcs))]); q = q * torch.exp(dro_eta * gl.detach()); q = q / q.sum(); loss = (q * gl).sum()
            elif meth == "CORAL": loss = ls.mean() + coral_w * coral_loss(f[d == 0], f[d == 1])
            else:                                                                    # DANN: 0.1 x domain loss, lambda ramps 0 -> 1
                p = (ep * spe + k) / (EPOCHS * spe); lam = 2 / (1 + math.exp(-10 * p)) - 1
                loss = ls.mean() + dann_w * F.cross_entropy(disc(GRL.apply(f, lam)), d)
            opt.zero_grad(set_to_none=True); scaler.scale(loss).backward(); scaler.step(opt); scaler.update(); sch.step(); tl += loss.detach()
        vs = val_scores(m, srcs); vm = float(np.mean(list(vs.values())))
        row = dict(epoch=ep + 1, train_loss=float(tl) / spe, val_mean=vm, **{f"val_{k}": v for k, v in vs.items()}, sec=time.time() - t0)
        hist.append(row); pd.DataFrame(hist).to_csv(HIST, index=False)
        print(f"ConvNeXt {ts} {meth} seed {seed} ep {ep + 1} | loss {row['train_loss']:.4f} | val {vs} | {row['sec'] / 60:.1f} min" + (f" | q={q.cpu().numpy().round(2)}" if meth == "GroupDRO" else ""))
        if vm > best: best = vm; torch.save(m.state_dict(), BEST)
        torch.save(dict(model=m.state_dict(), opt=opt.state_dict(), sched=sch.state_dict(), scaler=scaler.state_dict(), hist=hist, best=best, ep=ep, q=q.cpu(), **({"disc": disc.state_dict()} if disc else {})), LAST)
    finish_and_save(m, srcs, BEST, PRD); del m, opt, sch, scaler, disc; gc.collect(); torch.cuda.empty_cache()
```

# Part C2: one exact, fast, paired bootstrap engine for every statistic

**Why this cell exists.** Re-ranking scores in every bootstrap replicate is slow, which forced earlier runs down to 300 resamples and made Holm correction meaningless. `FastAUC` sorts the scores **once** and then computes the exact Mann-Whitney AUROC under integer sample weights (the number of times each image's patient was drawn), so 2,000 replicates are cheap.

**Pairing.** Every call to `run_boot(..., seed=11)` draws the *same* patient resamples in the *same order*. Therefore a difference between any two models, strata or analyses is a paired difference on identical patients, even when the two were computed in different cells.

**Definition used everywhere.** An AUROC reported for an architecture/training set is the **mean over seeds of the per-seed AUROC** (every number corresponds to a single trained model). Seed-ensemble AUROCs are used only for ensembles and in the sensitivity check of Part L1.

```python
class FastAUC:
    # Exact Mann-Whitney AUROC when image i carries the integer weight w[i] (how many times its patient was drawn). Scores are sorted once.
    def __init__(self, y, s):
        o = np.argsort(s, kind="mergesort"); ss = np.asarray(s)[o]
        self.o = o; self.y = np.asarray(y)[o].astype(float); self.starts = np.flatnonzero(np.r_[True, ss[1:] != ss[:-1]])      # tie groups
    def auc(self, w):
        w = w[self.o]; gp = np.add.reduceat(w * self.y, self.starts); gn = np.add.reduceat(w * (1 - self.y), self.starts)
        P_, N_ = gp.sum(), gn.sum()
        if P_ <= 0 or N_ <= 0: return np.nan
        return float((gp * (np.cumsum(gn) - gn + 0.5 * gn)).sum() / (P_ * N_))
def mkfa(P, T, min_n=5):                                           # one FastAUC per finding that has enough positives and negatives in test site T
    return {i: FastAUC(YT[T][:, i], P[:, i]) for i in range(L) if VALID[T][i] and min_n <= YT[T][:, i].sum() <= len(YT[T]) - min_n}
def boot_weights(nb, seed=11, unit="patient"):                     # same seed -> identical patient resamples in every call = paired across cells
    rng = np.random.default_rng(seed)
    for b in range(nb):
        yield b, {T: np.bincount(resample(T, rng) if unit == "patient" else rng.integers(0, len(YT[T]), len(YT[T])), minlength=len(YT[T])).astype(float) for T in "ABC"}
def run_boot(items, nb=None, seed=11, unit="patient", desc="paired bootstrap"):
    # items: {key: (site, {finding: FastAUC}, mask or None)}  ->  {key: (nb, L) AUROC per replicate}
    nb = NB if nb is None else nb; out = {k: np.full((nb, L), np.nan) for k in items}
    for b, w in tqdm(boot_weights(nb, seed, unit), total=nb, desc=desc):
        for k, (T, fl, mask) in items.items():
            ww = w[T] if mask is None else w[T] * mask
            for i, fa in fl.items(): out[k][b, i] = fa.auc(ww)
    return out
def point_auc(items):                                              # point estimates on the full test sets
    return {k: np.array([fl[i].auc(np.ones(len(YT[T])) if mask is None else np.asarray(mask, float)) if i in fl else np.nan for i in range(L)]) for k, (T, fl, mask) in items.items()}

_k = next(iter(PR3)); _s = next(iter(PR3[_k])); _T = next(iter(PR3[_k][_s])); _P = PR3[_k][_s][_T]                  # unit test against fauc on a real patient-level resample
_ix = resample(_T, np.random.default_rng(0)); _w = np.bincount(_ix, minlength=len(YT[_T])).astype(float)
for _i in range(L):
    if VALID[_T][_i]: assert abs(fauc(YT[_T][_ix, _i], _P[_ix, _i]) - FastAUC(YT[_T][:, _i], _P[:, _i]).auc(_w)) < 1e-9, _i
print(f"FastAUC agrees with the reference AUROC on a real resample | NB = {NB} | smallest attainable p = {1 / NB:.4f}")
```

# Part F: what explains the cross-hospital gap?

Three candidate explanations are examined:

1. **Case mix (view position).** Disease prevalence differs strongly by view (for example Edema is about 30% in CheXpert AP and 6% in CheXpert PA). If the asymmetry shrinks when only PA or only AP images are scored, case mix explains part of it. This is tested here for **all three architectures** with 2,000 paired resamples, three pre-defined Holm families and an effect size (the share of the asymmetry that remains inside each view). The DenseNet-only version with size-matched models is in Part L.
2. **Prevalence (label) shift.** Part G tests a label-free correction. It does not work here, and the reason is itself informative.
3. **Training-set size.** Part L: a size-matched control with repeated random CheXpert subsets.

These are attributions by conditioning, not a causal decomposition, and the terms can interact.

```python
if VIEW is None:
    print("view columns missing: this analysis is skipped")
else:
    VIEWS = ["PA", "AP"]
    OKV = {v: [j for j in range(L) if min(min(YT[T][VIEW[T] == v, j].sum(), (VIEW[T] == v).sum() - YT[T][VIEW[T] == v, j].sum()) for T in "AB") >= 30] for v in VIEWS}
    LABS_V = sorted(set(OKV["PA"]) & set(OKV["AP"]))                                  # the SAME findings in every stratum, so All, PA and AP are comparable
    print("findings used in the view analysis:", [LABELS6[j] for j in LABS_V])
    prv = pd.concat({f"{SITE[T]} ({T})": pd.DataFrame(YT[T], columns=LABELS6).groupby(VIEW[T]).mean() for T in "AB"}).round(3); display(prv); save_tab(prv, "table9_prevalence_by_view")

    CELLS_V = [("A", "A"), ("A", "B"), ("B", "B"), ("B", "A")]                         # in-hospital and external AUROC of the single-source models
    FAV = {(a, ts, T, s): mkfa(PR3[(a, ts, NAT)][s][T], T) for a in A3 for ts, T in CELLS_V for s in PR3[(a, ts, NAT)]}
    ITEMS_V = {}
    for (a, ts, T, s), fl in FAV.items():
        for st in ("All", "PA", "AP"):
            ITEMS_V[(a, ts, T, s, st)] = (T, {i: f for i, f in fl.items() if i in LABS_V}, None if st == "All" else (VIEW[T] == st).astype(float))
    OUTV = run_boot(ITEMS_V, NB, seed=11, desc="view-stratified paired bootstrap"); PTV = point_auc(ITEMS_V)
    def seedsV(a, ts): return list(PR3[(a, ts, NAT)])
    def gapV(a, st, d, store, point=False):                                            # in-hospital minus external, mean over seeds, macro over findings
        i_h, i_x = (("A", "A"), ("A", "B")) if d == "ab" else (("B", "B"), ("B", "A"))
        v = np.mean([store[(a, *i_h, s, st)] - store[(a, *i_x, s, st)] for s in seedsV(a, i_h[0])], 0)
        return np.nanmean(v) if point else np.nanmean(v[:, LABS_V], 1)
    Q = {}
    for a in A3:
        for st in ("All", "PA", "AP"):
            ab, ba = gapV(a, st, "ab", OUTV), gapV(a, st, "ba", OUTV); pab, pba = gapV(a, st, "ab", PTV, True), gapV(a, st, "ba", PTV, True)
            Q[(a, st, "gap A→B")] = (pab, ab); Q[(a, st, "gap B→A")] = (pba, ba); Q[(a, st, "asymmetry")] = (pab - pba, ab - ba)
    rows = []
    def addv(fam, a, q, st, est, d): lo, hi = np.percentile(d, [2.5, 97.5]); rows.append(dict(family=fam, arch=a, quantity=q, stratum=st, estimate=float(est), lo=lo, hi=hi, p=boot_p(d, NB)))
    for a in A3:
        for st in ("PA", "AP"):
            for q in ("gap A→B", "gap B→A"): addv("F1 view-matched gaps vs 0", a, q, st, *Q[(a, st, q)])                      # F1: 12 tests
            addv("F2 view-matched asymmetry vs 0", a, "asymmetry", st, *Q[(a, st, "asymmetry")])                                # F2: 6 tests
        for q in ("gap A→B", "gap B→A", "asymmetry"): addv("F3 PA − AP contrast", a, q, "PA − AP", Q[(a, "PA", q)][0] - Q[(a, "AP", q)][0], Q[(a, "PA", q)][1] - Q[(a, "AP", q)][1])   # F3: 9 tests
    VG = pd.DataFrame(rows); VG["p_holm"] = VG.groupby("family").p.transform(lambda x: holm(x.values)); VG["significant"] = VG.p_holm < .05
    print("Holm inside each family: F1 12 tests, F2 6 tests, F3 9 tests (families fixed in advance)"); display(VG.round(4)); save_tab(VG, "table10_view_stratified_robust", index=False)
    rows = []                                                                          # effect size: share of the all-image asymmetry that remains inside each view
    for a in A3:
        for st in ("PA", "AP"):
            den, num = Q[(a, "All", "asymmetry")][1], Q[(a, st, "asymmetry")][1]; ok = np.abs(den) > .01; rat = num[ok] / den[ok]
            rows.append(dict(arch=a, stratum=st, asymmetry_in_view=Q[(a, st, "asymmetry")][0], asymmetry_all=Q[(a, "All", "asymmetry")][0], share_remaining=Q[(a, st, "asymmetry")][0] / Q[(a, "All", "asymmetry")][0], lo=np.percentile(rat, 2.5), hi=np.percentile(rat, 97.5)))
    EFV = pd.DataFrame(rows); display(EFV.round(3)); save_tab(EFV, "table10b_view_effect_sizes", index=False)
    print("share_remaining near 1 inside BOTH views: case mix does not explain the asymmetry. Near 0: it does. In between: it explains part of it and the rest persists.")
    fig, axes = plt.subplots(1, 3, figsize=(COL2, 3.5))
    for a_, fam_ in zip(axes, sorted(VG.family.unique())):
        g_ = VG[VG.family == fam_].reset_index(drop=True)
        for i, r in g_.iterrows(): a_.errorbar(r.estimate, i, xerr=[[r.estimate - r.lo], [r.hi - r.estimate]], fmt=MKR[r.arch], color="k", mfc="k" if r.significant else "w", ms=4, capsize=2, lw=.8)
        a_.axvline(0, color="r", ls="--", lw=.7); a_.set_yticks(range(len(g_))); a_.set_yticklabels([f"{r.arch[:6]} {r.quantity.replace('gap ', '')} {r.stratum}" for r in g_.itertuples()], fontsize=4.6); a_.invert_yaxis()
        a_.set_title(fam_, fontsize=7); a_.set_xlabel("AUROC difference (95% CI)", fontsize=6); a_.grid(True, axis="x", alpha=.3)
    fig.suptitle("Filled = significant after Holm correction within the family", fontsize=7); fig.tight_layout(); save_fig(fig, "fig07_view_matched_robust")
```

# Part G: calibration, label-free prior-shift correction, k-label recalibration

* **G1** ECE, Brier score and calibration-in-the-large (mean predicted probability minus observed prevalence) for every external direction.
* **G2** The **k-label** correction: Platt scaling and a re-tuned F1 threshold fitted on k labelled target-validation images (k = 0 ... 5000, 20 random draws), for A→B and B→A in all three architectures. These corrections are monotone, so **AUROC cannot change**: they can repair calibration and thresholds, not ranking quality.
* **G3** A **label-free** correction (EM prior-shift adjustment, Saerens et al.) that needs no target labels. # Part H: domain generalization methods vs ERM

For each architecture, ERM (natural), ERM (balanced), DANN, CORAL and GroupDRO are compared on the **unseen** hospital of each pooled set (A+B → C, A+C → B, B+C → A).

**Two baselines are used, because the sampling scheme is itself a variable.** DANN, CORAL and GroupDRO all train with balanced sampling (equal images per hospital per step), while ERM (natural) samples proportionally to hospital size. A method beating ERM (natural) could therefore be winning because of balanced sampling alone. The **fair baseline for a DG method is ERM (balanced)**. Both are reported, each with its own Holm family.

```python
# ---------------- G3: label-free prior-shift correction (needs no target labels) ----------------
# Plain Saerens EM assumes calibrated source posteriors. Deep networks are not, and then EM can run away. Following Alexandari et al. (2020), the
# posteriors are first calibrated on the SOURCE validation set (Platt), then EM estimates the target prevalence. Three variants are compared.
def saerens(p, pi_s, iters=300, tol=1e-7):
    p = np.clip(p, 1e-6, 1 - 1e-6); pi_t = pi_s
    for _ in range(iters):
        num = p * (pi_t / pi_s); q = num / (num + (1 - p) * ((1 - pi_t) / (1 - pi_s))); new = float(np.clip(q.mean(), 1e-4, 1 - 1e-4))
        if abs(new - pi_t) < tol: break
        pi_t = new
    return q, pi_t
rows = []
for a in A3:
    for src in ["A", "B"]:
        T = "B" if src == "A" else "A"
        if (a, src, NAT) not in PR3: continue
        Y = YT[T]; Ysv = SRC[src]["Y"][SRC[src]["va"]]; prev_t = Y.mean(0)
        for s in PR3[(a, src, NAT)]:
            ex = extra_preds(a, src, s)
            if ex is None: continue
            P = PR3[(a, src, NAT)][s][T]; Q_raw, Q_cal, Q_em = P.copy(), P.copy(), P.copy(); est_raw, est_em = np.zeros(L), np.zeros(L)
            for j in range(L):
                pri_val = float(Ysv[:, j].mean()); a_, b_ = fit_platt(logit(ex["src_val"][:, j]), Ysv[:, j]); Pc = sig(a_ * logit(P[:, j]) + b_)
                Q_cal[:, j] = Pc                                                   # variant 2: calibrate on the source validation set only
                Q_raw[:, j], est_raw[j] = saerens(P[:, j], float(SRC[src]["Y"][SRC[src]["tr"]][:, j].mean()))      # variant 1: EM on uncalibrated posteriors
                Q_em[:, j], est_em[j] = saerens(Pc, pri_val)                       # variant 3: calibrate on source validation, then EM
            f = lambda Q: np.mean([ece_of(Y[:, j], Q[:, j]) for j in range(L)]); c = lambda Q: np.mean([Q[:, j].mean() - prev_t[j] for j in range(L)])
            rows.append(dict(arch=a, direction=dname(src, T), seed=s, ece_none=f(P), ece_em_uncal=f(Q_raw), ece_cal_only=f(Q_cal), ece_cal_em=f(Q_em),
                             citl_none=c(P), citl_cal_em=c(Q_em), prev_err_source=np.mean(np.abs(SRC[src]["Y"][SRC[src]["tr"]].mean(0) - prev_t)),
                             prev_err_em_uncal=np.mean(np.abs(est_raw - prev_t)), prev_err_cal_em=np.mean(np.abs(est_em - prev_t))))
SH = pd.DataFrame(rows)
if len(SH) == 0: print("no inputs for the label-free correction: see the G2 cell")
else:
    save_tab(SH, "table14_prior_shift_numeric", index=False)
    shs = SH.groupby(["direction", "arch"]).mean(numeric_only=True).round(4); display(shs); save_tab(shs, "table14_prior_shift_correction")
    print("ece_*: ECE after each variant. prev_err_*: mean absolute error of the assumed target prevalence (source prevalence vs the EM estimate).")
    print("AUROC is unchanged by all variants (monotone). If ece_em_uncal is WORSE than ece_none, plain EM failed because the posteriors are not calibrated.")
    worse = shs[shs.prev_err_cal_em > shs.prev_err_source]
    print(f"\nVERDICT: EM estimated the target prevalence WORSE than simply assuming no shift in {len(worse)} of {len(shs)} direction/architecture cells.")
    print("A label-shift correction only works if the shift is in the class prior and nothing else. Here it is not (case mix, labeling and image appearance differ too), so this label-free correction must not be claimed as a fix.")
    dirs_ = list(dict.fromkeys(SH.direction)); w = .2; fig, ax = plt.subplots(figsize=(COL2, 3.0))
    for i, (col, nm, hat) in enumerate([("ece_none", "none", ""), ("ece_em_uncal", "EM, uncalibrated", "\\\\"), ("ece_cal_only", "source-calibrated only", "xx"), ("ece_cal_em", "source-calibrated + EM", "///")]):
        for k, a in enumerate(A3):
            g_ = SH[SH.arch == a].groupby("direction")[col].mean().reindex(dirs_)
            ax.bar(np.arange(len(dirs_)) + (i - 1.5) * w + (k - 1) * w / 3.4, g_.values, w / 3.4, color=AC[a], hatch=hat, edgecolor="k", lw=.3, label=f"{nm}" if k == 0 else None)
    ax.set_xticks(range(len(dirs_))); ax.set_xticklabels(dirs_); ax.set(ylabel="ECE on the external site"); ax.legend(frameon=False, fontsize=6); ax.grid(True, axis="y", alpha=.3)
    ax.set_title("Label-free correction (within each group: DenseNet121, ConvNeXt-T, ViT-S)", fontsize=7); fig.tight_layout(); save_fig(fig, "fig11_prior_shift_correction")
```

```python
DGKEYS = [(a, ts, me) for a in A3 for ts in POOLED for me in METHS if (a, ts, me) in PR3]
REFK = list(dict.fromkeys((a, X, UNSEEN[ts]) for a in A3 for ts in POOLED for X in ts.split("+") if X != "C" and (a, X, NAT) in PR3))      # single-source references for the "pooling value" analysis
ITEMS_D = {("pool", a, ts, me, s): (UNSEEN[ts], mkfa(PR3[(a, ts, me)][s][UNSEEN[ts]], UNSEEN[ts]), None) for (a, ts, me) in DGKEYS for s in PR3[(a, ts, me)]}
ITEMS_D.update({("ref", a, X, T, s): (T, mkfa(PR3[(a, X, NAT)][s][T], T), None) for (a, X, T) in REFK for s in PR3[(a, X, NAT)]})
OUTD = run_boot(ITEMS_D, NB, seed=11, desc="DG bootstrap, per-seed AUROC"); PTD = point_auc(ITEMS_D)                                      # same seed as Part E: identical patients
mac = lambda x: np.nanmean(x, -1)
BQ = {(a, ts, me): np.mean([mac(OUTD[("pool", a, ts, me, s)]) for s in PR3[(a, ts, me)]], 0) for (a, ts, me) in DGKEYS}                  # (NB,)
PQ = {(a, ts, me): np.mean([np.nanmean(PTD[("pool", a, ts, me, s)]) for s in PR3[(a, ts, me)]]) for (a, ts, me) in DGKEYS}
BR = {(a, X, T): np.mean([mac(OUTD[("ref", a, X, T, s)]) for s in PR3[(a, X, NAT)]], 0) for (a, X, T) in REFK}
PR_ = {(a, X, T): np.mean([np.nanmean(PTD[("ref", a, X, T, s)]) for s in PR3[(a, X, NAT)]]) for (a, X, T) in REFK}
PSd = {k: True for k in BQ}; MARGIN = 0.01                                                                                              # practical-equivalence margin, fixed in advance: 0.01 AUROC

rows = []
for base, methods in [(NAT, METHS[1:]), ("ERM (balanced)", METHS[2:])]:
    fam = []
    for a in A3:
        for ts in POOLED:
            if (a, ts, base) not in PSd: continue
            for me in methods:
                if (a, ts, me) not in PSd: continue
                d = BQ[(a, ts, me)] - BQ[(a, ts, base)]
                lo_, hi_ = np.percentile(d, [2.5, 97.5]); fam.append(dict(baseline=base, arch=a, train=ts, unseen=UNSEEN[ts], method=me, diff=PQ[(a, ts, me)] - PQ[(a, ts, base)], lo=lo_, hi=hi_, p=boot_p(d, NB), mdd80=2.8 * d.std(), equivalent_within_margin=bool(lo_ > -MARGIN and hi_ < MARGIN)))
    fam = pd.DataFrame(fam); fam["p_holm"] = holm(fam.p.values); fam["significant"] = fam.p_holm < .05; rows.append(fam)
    print(f"baseline {base}: {len(fam)} comparisons, smallest attainable Holm p = {len(fam) / NB:.4f}")
DGS = pd.concat(rows, ignore_index=True); display(DGS.round(4)); save_tab(DGS, "table15_dg_vs_erm", index=False)
print("\nDG methods that BEAT their baseline after Holm correction:"); display(DGS[DGS.significant & (DGS["diff"] > 0)][["baseline", "arch", "train", "method", "diff", "lo", "hi", "p_holm"]].round(4))
print("DG methods that are significantly WORSE:", int((DGS.significant & (DGS["diff"] < 0)).sum()), "of", len(DGS))

rows = []
for (a, ts, me), d in PR3.items():
    if ts not in POOLED: continue
    T = UNSEEN[ts]; v = np.where(VALID[T])[0]
    for s, z in d.items():
        sa = [mauc(YT[t], z[t], VALID[t]) for t in SEEN_T[ts]]
        rows.append(dict(arch=a, train=ts, unseen=T, method=me, seed=s, auroc=mauc(YT[T], z[T], VALID[T]), auprc=mauprc(YT[T], z[T], VALID[T]),
                         ece=np.mean([ece_of(YT[T][:, j], z[T][:, j]) for j in v]), seen_mean=np.mean(sa), seen_worst=np.min(sa)))
DG3 = pd.DataFrame(rows); save_tab(DG3, "table16_dg_methods_numeric", index=False)
g = DG3.groupby(["arch", "train", "method"])
TABDG = pd.DataFrame({"Unseen AUROC": g.auroc.agg(lambda x: f"{x.mean():.3f} ± {x.std(ddof=1) if len(x) > 1 else 0:.3f}"), "Unseen AUPRC": g.auprc.agg(lambda x: f"{x.mean():.3f} ± {x.std(ddof=1) if len(x) > 1 else 0:.3f}"),
                      "Unseen ECE": g.ece.agg(lambda x: f"{x.mean():.3f} ± {x.std(ddof=1) if len(x) > 1 else 0:.3f}"), "Seen AUROC (mean)": g.seen_mean.agg(lambda x: f"{x.mean():.3f}"),
                      "Seen AUROC (worst)": g.seen_worst.agg(lambda x: f"{x.mean():.3f}"), "seeds": g.seed.nunique()})
display(TABDG.reindex(pd.MultiIndex.from_product([A3, SETS_ORDER, METHS], names=["arch", "train", "method"])).dropna(how="all")); save_tab(TABDG, "table16_dg_methods")
```

## H2: does pooling help at all? (and what the DG methods are really compared with)

A DG method is only interesting if it beats the simplest alternative. For the pooled sets that contain VinBigData, that alternative is to **train on the single large hospital only** and ignore the small, label-poor VinBigData source. This table compares every pooled model with the single-source model of each large hospital it contains, on the same unseen site, in one Holm family.

* `mdd80` = minimal detectable difference with 80% power (2.8 x the bootstrap SD of the paired difference). A non-significant result is only informative if `mdd80` is small.
* `equivalent_within_margin` = the whole 95% CI lies inside ±0.01 AUROC, a margin fixed in advance. That is positive evidence of no meaningful difference, which a non-significant p-value alone is not.

```python
class ViTCAM:                                                                          # Grad-CAM for ViT: last block norm1 tokens, patch tokens only
    def __init__(self, m):
        self.m = m; self.np_ = m.num_prefix_tokens; m.blocks[-1].norm1.register_forward_hook(self._fwd)
    def _fwd(self, mod, i, out):
        self.act = out
        if out.requires_grad: out.register_hook(lambda g: setattr(self, "grad", g))
    def __call__(self, x, j):                                                          # j: int or LongTensor [B] (finding per image)
        self.m.zero_grad(set_to_none=True)
        with torch.enable_grad():
            z = self.m(x); j = torch.full((len(x),), j, device=x.device) if isinstance(j, int) else j
            z.gather(1, j.view(-1, 1)).sum().backward()
        a_, g_ = self.act[:, self.np_:].detach(), self.grad[:, self.np_:].detach()
        cam = torch.relu((g_.mean(1, keepdim=True) * a_).sum(-1)); n = int(cam.shape[1] ** .5)
        return F.interpolate(cam.view(-1, 1, n, n), size=224, mode="bilinear", align_corners=False)[:, 0].cpu().numpy(), torch.sigmoid(z.detach()).cpu().numpy()
def make_cam(m, a):
    if a == "ViT-S": vc = ViTCAM(m); return lambda x, j: vc(x, j)[0]
    def f(x, j):                                                                       # conv Grad-CAM on the last feature map (DenseNet121, ConvNeXt-T)
        with torch.enable_grad():
            fm = m.forward_features(x); out = m.forward_head(fm); g_ = torch.autograd.grad(out[:, j].sum(), fm)[0]
        cam = torch.relu((g_.mean((2, 3), keepdim=True) * fm).sum(1, keepdim=True))
        return F.interpolate(cam, size=224, mode="bilinear", align_corners=False)[:, 0].detach().cpu().numpy()
    return f
nrm = lambda c: c / (c.reshape(len(c), -1).max(1)[:, None, None] + 1e-8)
def bp(ax, data, names):                                                               # boxplot that works on old and new Matplotlib
    try: ax.boxplot(data, tick_labels=names, showfliers=False)
    except TypeError: ax.boxplot(data, labels=names, showfliers=False)

# --- J2a: ViT-S true positives / false positives / false negatives on external sites ---
VCASES = [("A", "B", "A→B (NIH model on CheXpert)"), ("B", "A", "B→A (CheXpert model on NIH)"), ("A+B", "C", "A+B→C (pooled model on VinBigData)")]
for ts, T, ttl in VCASES:
    ck = stem("ViT-S", ts, NAT, 0) + ".pt"
    if not os.path.exists(ck): print("skipped (no checkpoint):", ck); continue
    m = load_arch("ViT-S", ck); cam = ViTCAM(m); A_, idxs = TEST[T]; P = PR3[("ViT-S", ts, NAT)][0][T]; Y = YT[T]
    for lab in ["Cardiomegaly", "Pleural Effusion"]:
        j = LABELS6.index(lab); p, y = P[:, j], Y[:, j]
        pick = {"True positive": np.where(y == 1)[0][np.argsort(-p[y == 1])[:4]], "False positive": np.where(y == 0)[0][np.argsort(-p[y == 0])[:4]], "False negative": np.where(y == 1)[0][np.argsort(p[y == 1])[:4]]}
        fig, ax = plt.subplots(3, 4, figsize=(COL2, 5.4))
        for r_, (nm, ii) in enumerate(pick.items()):
            cams, probs = cam(gpu_batch(A_, idxs[ii], a="ViT-S"), j)
            for c_ in range(4):
                a_ = ax[r_, c_]; a_.imshow(A_[idxs[ii[c_]]], cmap="gray"); a_.imshow(nrm(cams)[c_], cmap="jet", alpha=.4); a_.axis("off"); a_.set_title(f"{nm}\np={probs[c_, j]:.2f}", fontsize=6)
        fig.suptitle(f"{ttl} · {lab}", fontsize=8); fig.tight_layout(); save_fig(fig, f"fig21_gradcam_vit_{ts.replace('+', 'p')}_to_{T}_{lab.replace(' ', '')}")
    del m, cam; torch.cuda.empty_cache()

# --- J2b: same images, three architectures (NIH-trained models on CheXpert positives) ---
PA_ = {"DenseNet121": os.path.join(OUT, "nih_train", "densenet121_nih_e7_seed0.pt"), "ConvNeXt-T": os.path.join(OUT, "part34", "convnext_tiny_nih_e7_seed0.pt"), "ViT-S": stem("ViT-S", "A", NAT, 0) + ".pt"}
AV36 = [a for a in A3 if os.path.exists(PA_[a])]
for a in A3:
    if a not in AV36: print(f"SKIPPED {a}: checkpoint not found -> {PA_[a]}")
LAB = "Pleural Effusion"; jl = LABELS6.index(LAB); pos = np.where(YT["B"][:, jl] == 1)[0]; rng = np.random.default_rng(7); NST = 150; idx_st = np.sort(rng.choice(pos, NST, replace=False)); show = idx_st[:6]
CAMS = {}
for a in AV36:
    m = load_arch(a, PA_[a]); cam = make_cam(m, a); out = []
    for i in range(0, NST, 25): out.append(cam(gpu_batch(CX_T, idx_st[i:i + 25], a=a), jl))
    CAMS[a] = nrm(np.concatenate(out)); del m, cam; torch.cuda.empty_cache()
fig, ax = plt.subplots(len(AV36), 6, figsize=(COL2, 1.3 * len(AV36) + .3), squeeze=False)
for r_, a in enumerate(AV36):
    for c_ in range(6): ax[r_, c_].imshow(CX_T[show[c_]], cmap="gray"); ax[r_, c_].imshow(CAMS[a][c_], cmap="jet", alpha=.4); ax[r_, c_].axis("off")
    ax[r_, 0].set_title(a, fontsize=7, loc="left")
fig.suptitle(f"Grad-CAM, {LAB}: NIH-trained models on CheXpert positives (same images in every row)", fontsize=7); fig.tight_layout(); save_fig(fig, "fig22_gradcam_three_architectures")

def bshare(c, f=.1):
    k = int(224 * f); mk_ = np.ones((224, 224), bool); mk_[k:-k, k:-k] = False; return c[:, mk_].sum(1) / (c.reshape(len(c), -1).sum(1) + 1e-8)
down = lambda c: F.avg_pool2d(torch.from_numpy(c)[:, None], 4)[:, 0].numpy().reshape(len(c), -1)
def top_iou(c1, c2, q=80):
    a_, b_ = c1 >= np.percentile(c1, q, axis=1, keepdims=True), c2 >= np.percentile(c2, q, axis=1, keepdims=True); return (a_ & b_).sum(1) / ((a_ | b_).sum(1) + 1e-8)
BR3 = pd.DataFrame([dict(arch=a, border_share=x) for a in AV36 for x in bshare(CAMS[a])]); BRt = BR3.groupby("arch").border_share.agg(["mean", "std"]).round(3)
print("uniform CAM = 0.36 in the outer 10% frame"); display(BRt); save_tab(BRt, "table23_gradcam_border")
rows = []
for a, b in combinations(AV36, 2):
    da, db = down(CAMS[a]), down(CAMS[b]); pear = np.array([np.corrcoef(x, y)[0, 1] for x, y in zip(da, db)]); iou = top_iou(da, db)
    rows += [dict(pair=f"{a} vs {b}", metric="Pearson r", value=v) for v in pear] + [dict(pair=f"{a} vs {b}", metric="top-20% IoU", value=v) for v in iou]
AG = pd.DataFrame(rows); AGt = AG.groupby(["pair", "metric"]).value.agg(["mean", "std"]).round(3); display(AGt); save_tab(AGt, "table24_gradcam_agreement")
fig, ax = plt.subplots(1, 3, figsize=(COL2, 2.7))
bp(ax[0], [BR3[BR3.arch == a].border_share for a in AV36], [a.replace("-", "-\n") for a in AV36]); ax[0].axhline(.36, color="r", ls="--", lw=.7); ax[0].set(ylabel="CAM mass in outer 10% border", title="Border share")
for a_, met in zip(ax[1:], ["Pearson r", "top-20% IoU"]):
    g = AG[AG.metric == met]; ps_ = list(g.pair.unique()); bp(a_, [g[g.pair == p].value for p in ps_], [p.replace(" vs ", "\nvs\n") for p in ps_]); a_.set(title=f"Agreement ({met})")
for a_ in ax: a_.grid(True, axis="y", alpha=.3)
fig.tight_layout(); save_fig(fig, "fig23_gradcam_statistics")
```

# Part L: statistical robustness, size-matched control and source composition

Everything here uses the exact paired bootstrap of Part C2 (same patients in every comparison, `NB` resamples), Holm correction inside **families fixed in advance**, and effect sizes with 95% CIs.

| Section | Question | Models |
|---|---|---|
| L1 | Is the headline asymmetry robust to the AUROC definition, to dropping a seed, to dropping a finding, and to the resampling unit? | all three architectures |
| L2 | Does training-set size explain the asymmetry? (size-matched control, 8 random CheXpert-77k subsets, view-stratified) | DenseNet121 |
| L3 | How does the mix of source hospitals at a fixed training budget change performance on seen and unseen hospitals? | DenseNet121 |

```python
# ---------------- L1a: per-seed AUROC (primary definition) versus seed-ensemble AUROC (sensitivity) ----------------
HEAD = [(a, ts, T) for a in A3 for ts, T in [("A", "A"), ("A", "B"), ("B", "B"), ("B", "A")] if (a, ts, NAT) in PR3]
ITEMS_S = {(a, ts, T): (T, mkfa(PM3[(a, ts, T)], T), None) for (a, ts, T) in HEAD}
OUTS = run_boot(ITEMS_S, NB, seed=11, desc="seed-ensemble sensitivity"); PTS_ = point_auc(ITEMS_S)
rows = []
for a in A3:
    ab = OUTS[(a, "A", "A")] - OUTS[(a, "A", "B")]; ba = OUTS[(a, "B", "B")] - OUTS[(a, "B", "A")]
    d_ens = np.nanmean(ab - ba, 1); d_ps = np.nanmean(GB3[a]["asym"], 1)
    e_ens = float(np.nanmean((PTS_[(a, "A", "A")] - PTS_[(a, "A", "B")]) - (PTS_[(a, "B", "B")] - PTS_[(a, "B", "A")]))); e_ps = float(np.nanmean(GP3[a]["asym"])); dd = d_ens - d_ps
    rows.append(dict(arch=a, per_seed_asymmetry=e_ps, ps_lo=np.percentile(d_ps, 2.5), ps_hi=np.percentile(d_ps, 97.5), ensemble_asymmetry=e_ens, ens_lo=np.percentile(d_ens, 2.5), ens_hi=np.percentile(d_ens, 97.5),
                     ensemble_minus_per_seed=e_ens - e_ps, diff_lo=np.percentile(dd, 2.5), diff_hi=np.percentile(dd, 97.5)))
L1a = pd.DataFrame(rows); display(L1a.round(4)); save_tab(L1a, "table26a_definition_sensitivity", index=False)
print("Averaging seeds changes both the in-hospital and the external AUROC, so the seed-ensemble asymmetry can differ from the per-seed one (on your runs about 0.073 versus 0.081). Report ONE definition as primary (per-seed) and the other as a sensitivity analysis.")

# ---------------- L1b: leave-one-seed-out ----------------
def asym_seeds(a, seeds, store, point=False):
    f = (lambda k: store[k]) if point else (lambda k: store[k])
    ab = np.mean([store[(a, "A", "A", s)] - store[(a, "A", "B", s)] for s in seeds], 0); ba = np.mean([store[(a, "B", "B", s)] - store[(a, "B", "A", s)] for s in seeds], 0)
    return np.nanmean(ab - ba) if point else np.nanmean(ab - ba, 1)
rows = []
for a in A3:
    seeds_all = [s for s in PR3[(a, "A", NAT)] if s in PR3[(a, "B", NAT)]]
    for s0 in seeds_all:
        rest = [s for s in seeds_all if s != s0]; d = asym_seeds(a, rest, OUTB); rows.append(dict(arch=a, variant=f"leave seed {s0} out", estimate=float(asym_seeds(a, rest, PTS, True)), lo=np.percentile(d, 2.5), hi=np.percentile(d, 97.5)))
L1b = pd.DataFrame(rows); display(L1b.round(4)); save_tab(L1b, "table26b_leave_one_seed_out", index=False)

# ---------------- L1c: leave-one-finding-out (is the asymmetry carried by a single finding?) ----------------
rows = []
for a in A3:
    for j in [None] + list(range(L)):
        keep = [i for i in range(L) if i != j]; d = np.nanmean(GB3[a]["asym"][:, keep], 1); est = float(np.nanmean(GP3[a]["asym"][keep]))
        rows.append(dict(arch=a, variant="all six findings" if j is None else f"without {LABELS6[j]}", estimate=est, lo=np.percentile(d, 2.5), hi=np.percentile(d, 97.5)))
L1c = pd.DataFrame(rows); display(L1c.round(4)); save_tab(L1c, "table26c_leave_one_finding_out", index=False)
pf = pd.DataFrame({a: GP3[a]["asym"] for a in A3}, index=LABELS6).round(3); print("asymmetry per finding (A→B gap minus B→A gap):"); display(pf); save_tab(pf, "table26d_asymmetry_per_finding")
print("A finding with a NEGATIVE value goes against the headline direction (heterogeneity). Say so in the paper.")

# ---------------- L1d: resampling unit (patient versus image) ----------------
ITEMS_U = {k: v for k, v in ITEMS_E.items() if (k[1], k[2]) in [("A", "A"), ("A", "B"), ("B", "B"), ("B", "A")]}
OUTU = run_boot(ITEMS_U, NB, seed=11, unit="image", desc="image-level bootstrap (sensitivity)")
rows = []
for a in A3:
    seeds_all = [s for s in PR3[(a, "A", NAT)] if s in PR3[(a, "B", NAT)]]; d_img = asym_seeds(a, seeds_all, OUTU); d_pat = np.nanmean(GB3[a]["asym"], 1)
    rows.append(dict(arch=a, patient_level_width=np.ptp(np.percentile(d_pat, [2.5, 97.5])), image_level_width=np.ptp(np.percentile(d_img, [2.5, 97.5])), width_ratio_patient_over_image=np.ptp(np.percentile(d_pat, [2.5, 97.5])) / np.ptp(np.percentile(d_img, [2.5, 97.5]))))
L1d = pd.DataFrame(rows); display(L1d.round(4)); save_tab(L1d, "table26e_bootstrap_unit", index=False)
print("A ratio above 1 means ignoring that images of one patient are correlated would make the CIs too narrow (anti-conservative).")

lad = []
for a in A3:
    r_ = L1a[L1a.arch == a].iloc[0]; lad += [(a, "primary: per-seed, patient bootstrap", r_.per_seed_asymmetry, r_.ps_lo, r_.ps_hi), (a, "seed-ensemble AUROC", r_.ensemble_asymmetry, r_.ens_lo, r_.ens_hi)]
    lad += [(a, r.variant, r.estimate, r.lo, r.hi) for r in L1b[L1b.arch == a].itertuples()] + [(a, r.variant, r.estimate, r.lo, r.hi) for r in L1c[(L1c.arch == a) & (L1c.variant != "all six findings")].itertuples()]
fig, axes = plt.subplots(1, 3, figsize=(COL2, 4.2), sharex=True)
for a_, a in zip(axes, A3):
    g_ = [t for t in lad if t[0] == a]
    for i, (_, lab, est, lo, hi) in enumerate(g_): a_.errorbar(est, i, xerr=[[est - lo], [hi - est]], fmt="o", color="k", ms=3, capsize=1.5, lw=.7)
    a_.axvline(0, color="r", ls="--", lw=.6); a_.set_yticks(range(len(g_))); a_.set_yticklabels([t[1] for t in g_], fontsize=4.8); a_.invert_yaxis(); a_.set_title(a, fontsize=7); a_.set_xlabel("Asymmetry (AUROC)", fontsize=6); a_.grid(True, axis="x", alpha=.3)
fig.suptitle("Robustness of the headline asymmetry: definition, seeds, findings", fontsize=7); fig.tight_layout(); save_fig(fig, "fig24_asymmetry_robustness")
```

## L2: size-matched control (DenseNet121) with repeated random subsets

* **100% NIH = NIH-77k** (all 77,353 NIH training images, so it cannot be subsampled). **CheXpert-152k** is the full CheXpert training set. **CheXpert-77k** is a random **patient-level** subset of exactly 77,353 CheXpert images, drawn again for every seed and extended to 8 subsets, trained with the same steps, epochs and learning rate as NIH-77k.
* Decomposition: original asymmetry = size effect + size-controlled asymmetry (an identity). The CI resamples test patients **and** the random subset draws (nested bootstrap).
* **Holm families (fixed in advance).** S-F1: the three decomposition tests. S-V1: view-matched gaps vs 0 (6 tests). S-V2: view-matched asymmetry vs 0 (4 tests). S-V3: PA minus AP contrasts (5 tests).
* The 5 additional CheXpert-77k subsets (seeds 3 to 7) come from your `che44-nih77` notebook (`train_cx_subset`). If fewer than 2 subsets are on disk this section stops with a message.

## L3: source composition at a fixed training budget (DenseNet121)

Five configurations with the **same 77,353-image budget** (1,209 steps per epoch, 7 epochs): 100% NIH, 75/25, 50/50, 25/75 and 100% CheXpert. The endpoints are your NIH-77k and CheXpert-77k models; only the three mixtures are trained, with 3 seeds each. Every run draws a new random patient-level subset from each source. Test sets: NIH test, CheXpert test and **VinBigData as an unseen third hospital** (five findings).

**Holm families, fixed in advance:**

| Family | Question | Tests |
|---|---|---|
| C-F1 | NIH test: each configuration vs 100% NIH | 4 |
| C-F2 | CheXpert test: each configuration vs 100% CheXpert | 4 |
| C-F3 | VinBigData: each configuration vs 100% CheXpert (reference chosen in advance) | 4 |
| C-F3b | **exploratory, post hoc**: VinBigData, each configuration vs 100% NIH (added after C-F3 showed 100% CheXpert is the weakest endpoint on VinBigData) | 4 |
| C-F4 | mean of the two hospitals: each mixture vs each single source | 6 |
| C-F5 | trend: AUROC change per +25 points of CheXpert share, five outcomes | 5 |

```python
FRACS = [0.0, 0.25, 0.5, 0.75, 1.0]; NAMEF = {0.0: "100% NIH", 0.25: "75% NIH + 25% CheXpert", 0.5: "50% NIH + 50% CheXpert", 0.75: "25% NIH + 75% CheXpert", 1.0: "100% CheXpert"}
N_TOTAL = NT_DN                                                                      # the SAME 77,353-image budget in every configuration
print(pd.DataFrame({"configuration": NAMEF.values(), "NIH images": [N_TOTAL - round(f * N_TOTAL) for f in FRACS], "CheXpert images": [round(f * N_TOTAL) for f in FRACS]}, index=[f"{int(100 * f)}% CheXpert" for f in FRACS]).to_string())
def patient_subset(pool_idx, pats, n_target, r):                                    # random PATIENT-level subset of exactly n_target images
    if n_target <= 0: return np.array([], dtype=int)
    if n_target >= len(pool_idx): return np.sort(pool_idx)
    cnt = pd.Series(pats).value_counts().to_dict(); up = np.unique(pats); r.shuffle(up); chosen, tot = set(), 0
    for p in up:
        chosen.add(p); tot += cnt[p]
        if tot >= n_target: break
    idx = pool_idx[np.isin(pats, list(chosen))]
    return np.sort(r.permutation(idx)[:n_target])
def mix_tag(f, seed): return f"{DN_MODEL}_mix{int(round(100 * f)):03d}cx_n{N_TOTAL}_e{EPOCHS}_seed{seed}"
def endpoint_vb_cx77(s):                                                              # VinBigData predictions of the 100%-CheXpert endpoint, cached
    out = os.path.join(CTRL, f"{DN_MODEL}_endpoint_f100_seed{s}_vb.npy")
    if os.path.exists(out): return np.load(out)
    ck = os.path.join(CTRL, f"{DN_MODEL}_cx{NT_DN}_e{EPOCHS}_seed{s}.pt"); assert os.path.exists(ck), f"missing checkpoint {ck}"
    m = load_arch(DN, ck); P = predict(m, VB_A, np.arange(NVB), a=DN); np.save(out, P); del m; gc.collect(); torch.cuda.empty_cache(); return P
```

```python
# ---------------- train the three mixtures (skipped when the files already exist in controls) ----------------
def train_mix(f, seed):
    use_arch(DN); tag = mix_tag(f, seed); PRD = os.path.join(CTRL, tag + "_preds.npz")
    if os.path.exists(PRD): print(f"{tag}: finished, skipping"); return
    BEST, CKPT, HIST = (os.path.join(CTRL, tag + s) for s in (".pt", "_last.pt", "_history.csv")); os.makedirs(CTRL, exist_ok=True)
    n_cx = int(round(f * N_TOTAL)); n_nih = N_TOTAL - n_cx; r = np.random.default_rng(9000 + int(1000 * f) + seed)            # same subset draws as your che44-nih77 runs
    ni = patient_subset(tr_i, nih.patient_id.values[tr_i], n_nih, r); ci_ = patient_subset(cx_tr, cxm.patient_id.values[cx_tr], n_cx, r)
    pidx = np.r_[ni, ci_]; psrc = np.r_[np.zeros(len(ni), int), np.ones(len(ci_), int)]
    print(f"{tag}: {len(ni)} NIH + {len(ci_)} CheXpert images = {len(pidx)}")
    torch.manual_seed(seed); np.random.seed(seed)
    m = make_model(); o = torch.optim.AdamW(m.parameters(), lr=CUR["lr"], weight_decay=CUR["wd"]); spe = int(np.ceil(len(pidx) / BS))
    sc = torch.optim.lr_scheduler.CosineAnnealingLR(o, T_max=EPOCHS * spe); sl = torch.amp.GradScaler("cuda", enabled=(dev == "cuda")); hist, best, start = [], 0.0, 0
    if os.path.exists(CKPT):
        ck = torch.load(CKPT, map_location=dev, weights_only=False); m.load_state_dict(ck["model"]); o.load_state_dict(ck["opt"]); sc.load_state_dict(ck["sched"]); sl.load_state_dict(ck["scaler"]); hist, best, start = ck["hist"], ck["best"], ck["ep"] + 1
    allv = np.ones(L, bool)
    for ep in range(start, EPOCHS):
        perm = np.random.default_rng(seed * 100 + ep).permutation(len(pidx)); m.train(); t0 = time.time(); rl, n = torch.zeros((), device=dev), 0
        for k in tqdm(range(spe), desc=f"{int(100 * f)}% CheXpert, seed {seed}, epoch {ep + 1}/{EPOCHS}"):
            sel = perm[k * BS:(k + 1) * BS]; ii, ss = pidx[sel], psrc[sel]; xs, ys = [], []
            a_, b_ = np.sort(ii[ss == 0]), np.sort(ii[ss == 1])                       # a batch contains images of both sources in proportion to the mixture
            if len(a_): xs.append(gpu_batch(NIH_A, a_, flip=True)); ys.append(Yn[a_])
            if len(b_): xs.append(gpu_batch(CX_A, b_, flip=True)); ys.append(Ycm[b_])
            x = torch.cat(xs); y = torch.from_numpy(np.concatenate(ys)).to(dev).float(); o.zero_grad(set_to_none=True)
            with torch.autocast("cuda", enabled=(dev == "cuda")): loss = F.binary_cross_entropy_with_logits(m(x).float(), y)
            sl.scale(loss).backward(); sl.step(o); sl.update(); sc.step(); rl += loss.detach() * len(x); n += len(x)
        vals = []                                                                     # selection on the validation sets of the sources that are present, never on a test set
        if len(ni): vals.append(mauc(Yn[va_i], predict(m, NIH_A, va_i), allv))
        if len(ci_): vals.append(mauc(Ycm[cx_va], predict(m, CX_A, cx_va), allv))
        vm = float(np.mean(vals)); row = dict(epoch=ep + 1, train_loss=float(rl) / n, val_auroc=vm, sec=time.time() - t0); hist.append(row); pd.DataFrame(hist).to_csv(HIST, index=False)
        print(f"{tag} ep {ep + 1} | loss {row['train_loss']:.4f} | val AUROC {vm:.4f} | {row['sec'] / 60:.1f} min")
        if vm > best: best = vm; torch.save(m.state_dict(), BEST)
        torch.save(dict(model=m.state_dict(), opt=o.state_dict(), sched=sc.state_dict(), scaler=sl.state_dict(), hist=hist, best=best, ep=ep), CKPT)
    m.load_state_dict(torch.load(BEST, map_location=dev, weights_only=True))
    np.savez(PRD, cx_test_P=predict(m, CX_T, np.arange(len(CX_T))), nih_test_P=predict(m, NIH_A, te_i), vb_P=predict(m, VB_A, np.arange(NVB))); del m, o, sc, sl; gc.collect(); torch.cuda.empty_cache()
RUN_MIX = True
if RUN_MIX:
    for f in (0.25, 0.5, 0.75):
        for s in SEEDS: train_mix(f, s)
```

# Part M: clinically meaningful operating points

AUROC says nothing about whether a model keeps a usable sensitivity and specificity under shift. For the single-source models (A→B and B→A, three architectures, three seeds) the threshold of every finding is **chosen on the source validation set only** at 90% specificity (or 90% sensitivity) and applied unchanged to the source test set and to the external hospital. **No target label is used.** Sensitivity, specificity and positive predictive value (PPV) come with 95% paired patient-bootstrap CIs. Do not mix these numbers with the k-label recalibration experiments of Part G, which use target labels.

**Holm family (fixed in advance):** the external-minus-in-hospital change in sensitivity and in specificity, 3 architectures x 2 directions x 2 operating modes x 2 metrics = 24 tests.

```python
NB_OP = NB; OPS = (("90% specificity", .90), ("90% sensitivity", .90)); COMB = []
for a in A3:
    for src in ["A", "B"]:
        T = "B" if src == "A" else "A"
        if (a, src, NAT) not in PR3: continue
        for s in PR3[(a, src, NAT)]:
            ex = extra_preds(a, src, s)
            if ex is None: continue
            Ysv = SRC[src]["Y"][SRC[src]["va"]]
            for j in range(L):
                neg, pos = ex["src_val"][Ysv[:, j] == 0, j], ex["src_val"][Ysv[:, j] == 1, j]
                if len(pos) < 20: continue
                for mode, lvl in OPS:
                    thr = np.quantile(neg, lvl) if mode == "90% specificity" else np.quantile(pos, 1 - lvl)               # threshold from SOURCE validation data only
                    for kind, site in (("in-hospital", src), ("external", T)):
                        P = PR3[(a, src, NAT)][s][site][:, j]; y = YT[site][:, j]; pred = (P > thr) if mode == "90% specificity" else (P >= thr)
                        COMB.append(dict(arch=a, src=src, seed=s, j=j, mode=mode, kind=kind, site=site, tp=(y * pred), fp=((1 - y) * pred), pos=y, neg=(1 - y)))
print(len(COMB), "threshold/site combinations")
if len(COMB) == 0: print("no operating-point inputs: check extra_preds / checkpoint paths in Part G")
else:
    BYS = {T: [c for c in COMB if c["site"] == T] for T in "AB"}
    MAT = {T: {k: np.stack([c[k] for c in BYS[T]]).astype(np.float32) for k in ("tp", "fp", "pos", "neg")} for T in "AB"}
    def ops_from(T, w): wf = w.astype(np.float32); tp, fp, P_, N_ = (MAT[T][k] @ wf for k in ("tp", "fp", "pos", "neg")); return np.stack([tp / P_, 1 - fp / N_, tp / np.maximum(tp + fp, 1e-9)], 1)       # sens, spec, PPV
    OPB = {T: np.zeros((NB_OP, len(BYS[T]), 3), np.float32) for T in "AB"}
    for b, w in tqdm(boot_weights(NB_OP, 11), total=NB_OP, desc="operating-point bootstrap"):
        for T in "AB": OPB[T][b] = ops_from(T, w[T])
    OPP = {T: ops_from(T, np.ones(len(YT[T]))) for T in "AB"}
    META = pd.DataFrame([{k: c[k] for k in ("arch", "src", "seed", "j", "mode", "kind", "site")} for c in COMB]); META["T"] = META.site; META["idx"] = META.groupby("site").cumcount()
    def agg(sel, m):                                                                # mean over seeds and findings -> (NB,), point
        d = None; pt = []; parts = []
        for T in "AB":
            q = sel[sel.site == T]
            if len(q): parts.append(np.nanmean(OPB[T][:, q.idx.values, m], 1)); pt.append(np.nanmean(OPP[T][q.idx.values, m]))
        return parts[0], pt[0]
    rows = []
    for a in A3:
        for src in ["A", "B"]:
            for mode, _ in OPS:
                for m, mn in enumerate(["sensitivity", "specificity", "PPV"]):
                    sh = META[(META.arch == a) & (META.src == src) & (META["mode"] == mode)]
                    if sh.empty: continue
                    bi, pi = agg(sh[sh.kind == "in-hospital"], m); be, pe = agg(sh[sh.kind == "external"], m); d = be - bi; lo, hi = np.percentile(d, [2.5, 97.5])
                    rows.append(dict(arch=a, direction="A→B" if src == "A" else "B→A", mode=mode, metric=mn, in_hospital=pi, in_lo=np.percentile(bi, 2.5), in_hi=np.percentile(bi, 97.5), external=pe, ext_lo=np.percentile(be, 2.5), ext_hi=np.percentile(be, 97.5), change=pe - pi, lo=lo, hi=hi, p=boot_p(d, NB_OP)))
    OP = pd.DataFrame(rows); fam = OP.metric.isin(["sensitivity", "specificity"]); OP.loc[fam, "p_holm"] = holm(OP.loc[fam, "p"].values); OP["significant"] = OP.p_holm < .05
    print(f"Holm family: {int(fam.sum())} tests (change in sensitivity and specificity); smallest attainable Holm p = {int(fam.sum()) / NB_OP:.4f}"); display(OP.round(3)); save_tab(OP, "table31_operating_points", index=False)
    print("The threshold is chosen for 90% specificity (or sensitivity) in the SOURCE hospital. A large change in the other quantity at the external hospital is the clinical cost of the shift. PPV also changes with prevalence.")
    fig, axes = plt.subplots(2, 2, figsize=(COL2, 4.8), sharey="row")
    for r_, mode in enumerate([o[0] for o in OPS]):
        for c_, mn in enumerate(["sensitivity", "specificity"]):
            a_ = axes[r_, c_]; g_ = OP[(OP["mode"] == mode) & (OP.metric == mn)]; labs = [(a, d_) for a in A3 for d_ in ("A→B", "B→A")]
            for k, (a, d_) in enumerate(labs):
                q = g_[(g_.arch == a) & (g_.direction == d_)]
                if q.empty: continue
                q = q.iloc[0]; a_.bar(k - .2, q.in_hospital, .38, yerr=[[q.in_hospital - q.in_lo], [q.in_hi - q.in_hospital]], capsize=1.5, color="0.8", edgecolor="k", lw=.4, label="in-hospital" if k == 0 else None)
                a_.bar(k + .2, q.external, .38, yerr=[[q.external - q.ext_lo], [q.ext_hi - q.external]], capsize=1.5, color="0.35", edgecolor="k", lw=.4, label="external" if k == 0 else None)
            a_.set_xticks(range(len(labs))); a_.set_xticklabels([f"{a[:5]}\n{d_}" for a, d_ in labs], fontsize=5.5); a_.set_title(f"{mn.capitalize()} at {mode} (threshold from source validation)", fontsize=6.5); a_.grid(True, axis="y", alpha=.3)
    axes[0, 0].legend(frameon=False, fontsize=6); fig.tight_layout(); save_fig(fig, "fig29_operating_points")
    rows = []                                                                       # per finding, external, 90%-specificity threshold
    for a in A3:
        for src in ["A", "B"]:
            for j in range(L):
                sh = META[(META.arch == a) & (META.src == src) & (META["mode"] == "90% specificity") & (META.kind == "external") & (META.j == j)]
                if sh.empty: continue
                T = sh.site.iloc[0]; ix_ = sh.idx.values; se = np.nanmean(OPB[T][:, ix_, 0], 1); sp = np.nanmean(OPB[T][:, ix_, 1], 1)
                rows.append(dict(arch=a, direction="A→B" if src == "A" else "B→A", finding=LABELS6[j], sensitivity=float(np.nanmean(OPP[T][ix_, 0])), sens_lo=np.percentile(se, 2.5), sens_hi=np.percentile(se, 97.5), specificity=float(np.nanmean(OPP[T][ix_, 1])), spec_lo=np.percentile(sp, 2.5), spec_hi=np.percentile(sp, 97.5)))
    OPF = pd.DataFrame(rows); display(OPF.round(3).head(12)); save_tab(OPF, "table32_operating_points_per_finding", index=False)
```

## Methods text (edit the bracketed parts)

*Data.* We used NIH ChestX-ray14 (A; official test list, patient-level train/validation split), CheXpert (B; patient-level split, uncertain labels treated as [your policy]) and VinBigData, built from VinDr-CXR (C; 15,000 images, a finding positive when at least 2 of 3 radiologists marked it). Six findings were studied (Cardiomegaly, Edema, Consolidation, Atelectasis, Pneumothorax, Pleural Effusion); C has no Edema, so every metric involving C uses the other five and its loss masks Edema. Images were resized to 224 x 224.

*Models.* DenseNet121, ConvNeXt-T and ViT-S (ImageNet pretraining) were fine-tuned with binary cross-entropy for 7 epochs, three seeds, each with its standard recipe (Part B). Models were trained on A, B, C, A+B, A+C and B+C. The checkpoint with the highest mean source-validation AUROC was kept; no target data were used for selection. DG methods (DANN, weight 0.1; CORAL, weight 1.0; GroupDRO, eta 0.01) and ERM with balanced sampling were run on the pooled sets.

*Statistics.* Macro AUROC over findings present in the test site, computed per seed and averaged over seeds. Confidence intervals and p-values use a paired bootstrap that resamples **test patients** (images for C, which has no patient identifier) [NB] times; the same sampled patients are used for every model and every analysis, and training-set draws and seeds are resampled where they are random. Two-sided bootstrap p-values are Holm-corrected within families fixed in advance (Table 33). The gap is in-hospital minus external AUROC; asymmetry is the NIH→CheXpert gap minus the CheXpert→NIH gap. Effect sizes are AUROC differences with 95% CIs. A practical-equivalence margin of 0.01 AUROC was fixed in advance. Operating points use thresholds chosen on source validation data only.

## Limitations to state in the paper

* Only two hospital sources with patient identifiers plus one easier site; no MIMIC-CXR or PadChest.
* VinBigData has no Edema and no patient identifier; comparisons that involve C use five findings and an image-level bootstrap.
* Images are 224 x 224; behaviour at higher resolution was not tested.
* Each architecture used its standard recipe; the three recipes are not identical, and no per-architecture hyperparameter search was run.
* **GroupDRO collapsed onto one hospital in every run (Table 17), so no domain-generalization claim is made about it.** The equal-budget tuning study (`REFER2`) is single-seed and its CORAL runs did not train, so it must not be cited as evidence about CORAL.
* DenseNet121 GroupDRO on the pooled sets with VinBigData predictions is not available. The size-matched control and the composition ablation are DenseNet121 only.
* The label-free prior-shift correction fails because the shift is not a pure prior shift; it is reported as a negative result.
* Grad-CAM is descriptive.
* The per-finding results are heterogeneous (at least one finding, Pleural Effusion in the size-matched analysis, goes against the headline direction).
* CheXpert labels come from an automatic labeler and NIH labels from report mining, so label noise differs by site.
